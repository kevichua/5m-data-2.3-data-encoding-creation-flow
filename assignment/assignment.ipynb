### Question 1

Question: Implement a simple Thrift server and client that defines a `Student` struct with fields `name` (string), `age` (integer), and `courses` (list of strings). Include a service `School` with a method `enrollCourse` that takes a `Student` record and a course name, adds the course to the student's course list, and returns the updated `Student` record.

Answer:

**Thrift schema (student.thrift)**

```python
%%writefile student.thrift
struct Student {
    1: string name,
    2: i32 age,
    3: list<string> courses
}

service School {
    Student enrollCourse(1: Student student, 2: string courseName)
}

```

**Thrift server (student_server.py)**

```python
%%writefile student_thrift_server.py

import thriftpy2
student_thrift = thriftpy2.load("student.thrift", module_name="student_thrift")

from thriftpy2.rpc import make_server

class School(object):
    def enrollCourse(self, student, courseName):
        """Add a course to the student's course list and return updated student"""
        print(f"📚 Enrolling {student.name} (Age: {student.age}) in: {courseName}")
        
        # Initialize courses list if None
        if student.courses is None:
            student.courses = []
        
        # Add course if not already enrolled
        if courseName not in student.courses:
            student.courses.append(courseName)
            print(f"✅ Added {courseName} to {student.name}'s courses")
            print(f"📖 Current courses: {', '.join(student.courses)}")
        else:
            print(f"⚠️ {student.name} is already enrolled in {courseName}")
        
        return student

# Create and start the server
server = make_server(student_thrift.School, School(), host='localhost', port=9090)
print("=" * 50)
print("🏫 School Thrift Server started (using thriftpy2)")
print(f"📍 Listening on: localhost:9090")
print("🛑 Press Ctrl+C to stop the server")
print("=" * 50)

server.serve()
```

```python
import thriftpy2
student_thrift = thriftpy2.load("student.thrift", module_name="student_thrift")

from thriftpy2.rpc import make_client

def print_student_info(student, title="Student Info"):
    """Helper function to display student information"""
    print(f"\n{'='*40}")
    print(f"📋 {title}")
    print(f"{'='*40}")
    print(f"👤 Name: {student.name}")
    print(f"🎂 Age: {student.age}")
    print(f"📚 Courses: {', '.join(student.courses) if student.courses else 'None'}")
    print(f"{'='*40}")

# Connect to the server
print("🚀 Connecting to Thrift server...")
school = make_client(student_thrift.School, host='localhost', port=9090)
print("✅ Connected successfully!\n")

# Example 1: Create a student with existing courses
student1 = student_thrift.Student(
    name="Alice Johnson",
    age=20,
    courses=["Mathematics", "Physics"]
)

print_student_info(student1, "Initial Student Record")

# Enroll in a new course
student1 = school.enrollCourse(student1, "Computer Science")
print_student_info(student1, "After Enrolling in Computer Science")

# Try enrolling in a duplicate course
student1 = school.enrollCourse(student1, "Physics")

# Example 2: Create a student with no courses
student2 = student_thrift.Student(
    name="Bob Smith",
    age=22,
    courses=[]
)

print_student_info(student2, "New Student (No Courses)")

# Enroll in multiple courses
for course in ["History", "Economics", "Political Science"]:
    student2 = school.enrollCourse(student2, course)

print_student_info(student2, "Final Course List")

# Example 3: Create another student
student3 = student_thrift.Student(
    name="Charlie Brown",
    age=19,
    courses=["Art", "Music"]
)

print_student_info(student3, "Another Student")

# Enroll in a course
student3 = school.enrollCourse(student3, "Theater")
print_student_info(student3, "After Theater Enrollment")

print("\n✨ All operations completed successfully!")
```

Run the thrift server

**Thrift client (student_client.py)**

```python
# Enter your student_client code here
import sys
sys.path.append('gen-py')

from student import School
from student.ttypes import Student

from thrift.transport import TSocket, TTransport
from thrift.protocol import TBinaryProtocol


def main():
    transport = TSocket.TSocket('localhost', 9090)
    transport = TTransport.TBufferedTransport(transport)
    protocol = TBinaryProtocol.TBinaryProtocol(transport)

    client = School.Client(protocol)
    transport.open()

    student = Student(name="Alice", age=20, courses=[])
    updated_student = client.enrollCourse(student, "Math")

    print("Updated Student:", updated_student)

    transport.close()


if __name__ == "__main__":
    main()
```
